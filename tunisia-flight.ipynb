## 7. OUTLIER AND ANOMALY ANALYSIS

One of the notable observations from the descriptive analysis was the maximum arrival delay of 3,451, compared with a median of 14.

This indicates that the dataset may contain extreme observations.

Using the interquartile range method:

- First quartile (Q1) = 0
- Third quartile (Q3) = 43
- Interquartile range (IQR) = 43

The upper outlier boundary is calculated as:

Upper boundary = Q3 + (1.5 × IQR)

Upper boundary = 43 + (1.5 × 43)

Upper boundary = 107.5

Therefore, recorded arrival delays above 107.5 would be flagged as potential upper-end outliers using this method.

These values should not automatically be removed because they may represent genuine operational delays.

Instead, they should be investigated to determine whether they result from genuine disruptions, recording errors or unusual circumstances.

The high maximum value and standard deviation indicate that further outlier analysis would be beneficial before performing predictive modelling.

## 9. RECOMMENDATIONS AND NEXT STEPS

Based on the findings, the following recommendations are proposed:

**1. Investigate extreme delay values**

Further examination should be conducted on observations with unusually high arrival delays to establish whether they are genuine or incorrectly recorded.

**2. Investigate zero arrival delays**

Zero values should be examined to distinguish flights that genuinely experienced no delay from observations where delay information may not have been recorded.

**3. Convert date and time variables**

Flight_date should be converted into a datetime format, while scheduled flight times should be appropriately processed.

This would allow further investigation into potential daily, weekly, monthly and seasonal delay patterns.

**4. Examine airport performance using sufficient observations**

Future airport comparisons should consider both average arrival delays and the number of flights recorded at each location.

Locations with very small samples should be treated cautiously.

**5. Investigate aircraft-related patterns**

The Aircraft_code variable could be used to compare average delays across different aircraft types, potentially identifying differences in operational performance.

**6. Conduct further statistical analysis**

Additional numerical features could be created from existing information, including departure hour, arrival hour, day of the week and month.

These features would support more meaningful correlation analysis.

**7. Explore predictive modelling**

Once the dataset has been cleaned and additional features have been prepared, machine learning techniques could be explored to investigate whether arrival delays can be predicted using available flight information.
